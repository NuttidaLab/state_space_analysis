```python
import matplotlib.pyplot as plt
import numpy as np



# Function to plot LBA model schematic
def plot_lba_schematic(params):
    s = params['s']
    tau = params['tau']
    b = params['b']
    A = params['A']
    v = params['v']

    # Plot for each condition
    n_conditions = len(v)
    fig, axes = plt.subplots(1, n_conditions, figsize=(15, 5), sharey=True)

    for i, ax in enumerate(axes):
        drift_rate = v[i]

        # Plotting
        ax.axhline(y=b, color='k', linestyle='--', label='Response threshold (b)')
        ax.axhline(y=0, color='k', linewidth=0.5)
        ax.fill_betweenx([0, b], 0, A, color='gray', alpha=0.3, label='Starting point range [0, A]')
        ax.arrow(A/2, 0, drift_rate, b, head_width=0.2, head_length=0.1, fc='black', ec='black', label='Drift rate')
        
        # Adding labels and titles
        ax.set_xlim(-1, A + 1)
        ax.set_ylim(-1, b + 1)
        ax.set_title(f'Target {i + 1}')
        ax.set_xlabel('Decision Time')
        if i == 0:
            ax.set_ylabel('Activation')

        ax.legend(loc='upper left', fontsize=8)

    plt.tight_layout()
    plt.show()

# Parameters
params = {
    's': 1,
    'tau': 0,
    'b': 1,
    'A': 0.2,
    'v': np.array([0.65924304, 0.74971889, 0.20725725, 0.37457809, 0.62086764])
}

# Plot the LBA model schematic
plot_lba_schematic(params)
```
